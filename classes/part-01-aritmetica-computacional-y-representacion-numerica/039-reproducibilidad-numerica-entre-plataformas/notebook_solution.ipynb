Solución ejecutable. Compárala con la tuya **después** de intentarlo.

```python
import json, sys
from pathlib import Path
ROOT = Path.cwd()
while ROOT != ROOT.parent and not (ROOT / 'curriculum.yaml').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
from computational_math.engines import part01 as motor
DEMO = 'reproducibility'
demo = motor.DEMOS[DEMO]
resultado = demo()
assert isinstance(resultado, dict) and resultado
assert len(resultado) == 7, 'la demostración cambió de forma'
print(json.dumps(resultado, indent=2, ensure_ascii=False, default=str))
```

## Interpretación de las salidas

`reproducibility` — El orden de la suma cambia el resultado en punto flotante.

Las claves booleanas son comprobaciones de invariante: si alguna fuera `False`,
el resultado numérico no sería fiable aunque el código se ejecutase sin error.


```python
invariantes = {k: v for k, v in resultado.items() if isinstance(v, bool)}
numericos = {k: v for k, v in resultado.items() if isinstance(v, (int, float))}
print('invariantes comprobados:', len(invariantes))
for k, v in invariantes.items():
    print(f'  {"OK " if v else "REVISAR"} {k}')
print('valores numéricos:', len(numericos))
```

```python
# Reproducibilidad: la demostración es determinista.
assert demo() == resultado, 'la demostración debe ser reproducible'
print('Reproducible: dos ejecuciones devuelven exactamente lo mismo.')
```

## Idea que hay que llevarse

> La cancelación catastrófica destruye dígitos significativos sin lanzar excepciones.

**Error a evitar:** Comparar floats con `==` en lugar de una tolerancia razonada.
