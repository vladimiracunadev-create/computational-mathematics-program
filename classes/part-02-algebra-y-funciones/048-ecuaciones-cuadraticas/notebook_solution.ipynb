## Interpretación de las salidas

`quadratic_equation` — Resolver una cuadrática y contrastar con la forma de vértice.

Las claves booleanas son comprobaciones de invariante: si alguna fuera `False`,
el resultado numérico no sería fiable aunque el código se ejecutase sin error.


```python
invariantes = {k: v for k, v in resultado.items() if isinstance(v, bool)}
numericos = {k: v for k, v in resultado.items() if isinstance(v, (int, float))}
print('invariantes comprobados:', len(invariantes))
for k, v in invariantes.items():
    print(f'  {"OK " if v else "REVISAR"} {k}')
print('valores numéricos:', len(numericos))
```

```python
# Reproducibilidad: la demostración es determinista.
assert demo() == resultado, 'la demostración debe ser reproducible'
print('Reproducible: dos ejecuciones devuelven exactamente lo mismo.')
```
